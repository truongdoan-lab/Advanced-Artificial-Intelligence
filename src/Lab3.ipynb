## 1. Introduction to Regular Expression (RegEx)

### 1.1. Theory: Basic Patterns and Operators
**Regular Expressions** (often shortened to **regex**) are an algebraic notation used for characterizing a set of strings. They are essential for searching through a **corpus** (a single document or collection) to find specific text patterns.

**Core Concepts:**
*   **Concatenation:** The simplest regex is a sequence of characters. For example, the pattern `woodchuck` matches any string containing that exact substring.
*   **Disjunction:** Square brackets `[ ]` allow you to match any single character from a set. For example, `/[wW]oodchuck/` matches either "Woodchuck" or "woodchuck".
*   **Ranges:** To avoid listing every character, you can use a dash `-`. For instance, `//` specifies any single digit, and `/[A-Z]/` matches any uppercase letter.
*   **Negation:** If the caret `^` is the first symbol after the open square bracket `[`, the pattern is negated. `/[^a-z]/` matches any single character that is **not** a lowercase letter.
*   **The Wildcard:** The period `.` matches any single character except a carriage return.


### 1.2. Python Sample: The `re` Module and Basic Matching
In Python, we use the `re` module to work with regular expressions. The `re.findall()` function is particularly useful for extracting all substrings that match a specific pattern.

```python
import re

# Sample corpus
text = "The woodchuck and the Woodchuck are both types of groundhogs. One costs $199 and the other $24.99."

# Example 1: Using Disjunction to find 'woodchuck' regardless of capitalization
pattern1 = r"[wW]oodchuck"
matches1 = re.findall(pattern1, text)
print(f"Matches for woodchuck: {matches1}")

# Example 2: Using Ranges to find all digits
pattern2 = r""
matches2 = re.findall(pattern2, text)
print(f"All individual digits found: {matches2}")

# Example 3: Using the Wildcard to find three-letter words starting with 't'
pattern3 = r"t.."
matches3 = re.findall(pattern3, text)
print(f"Three-character sequences starting with 't': {matches3}")
```

## 2. Advanced Regex—Groups, Substitutions, and Complexity


### 2.1 Theory: Grouping and Precedence
In complex searches, you often need to group patterns together or specify which operator should apply first.

**Key Concepts:**
*   **Disjunction (`|`):** The pipe symbol acts as an "OR" operator. For example, `/cat|dog/` matches either "cat" or "dog".
*   **Grouping with Parentheses `( )`:** Parentheses are used to group terms so that operators like the Kleene star or the pipe apply to the whole group. For instance, `/gupp(y|ies)/` ensures the disjunction applies only to the suffixes "y" and "ies".
*   **Operator Precedence:** Regular expressions follow a hierarchy from highest to lowest precedence:
    1.  **Parentheses:** `( )`
    2.  **Counters (Quantifiers):** `* + ? { }`
    3.  **Sequences and Anchors:** `the ^my end$`
    4.  **Disjunction:** `|`.
*   **Greedy vs. Non-Greedy:** By default, regex patterns are **greedy**, meaning they match the largest possible string. To make them **non-greedy** (matching as little as possible), use a question mark after the quantifier: `*?` or `+?`.

### 2.2 Python Sample: Capture Groups and Substitutions
**Capture groups** store matched patterns in memory registers (numbered `\1`, `\2`, etc.), allowing you to refer back to them during substitutions. In Python, we use `re.sub()` for these transformations.


```python
import re

# Sample: Surrounding all integers in a text with angle brackets < >
text = "There are 35 boxes and 12 crates."

# The parenthesis (+) creates a capture group
# \1 in the replacement string refers to the content of that first group
processed_text = re.sub(r"(+)", r"<\1>", text)

print(f"Original: {text}")
print(f"Substituted: {processed_text}")
# Output: There are <35> boxes and <12> crates.
```

In NLP, we often struggle with two types of errors:

## 3. Text Normalization and Corpora Analysis

### 3.1 Theory: Tokenization and Normalization
Before any NLP task, text must be converted into a standard form. This process involves several layers:

**Key Concepts:**
*   **Tokenization:** The task of separating a running stream of text into individual units called **tokens**. While whitespace is a common delimiter, it is often insufficient for cases like "New York" (a single entity with a space) or clitics like "I'm" (which should be "I" and "am").
*   **Normalization:** Mapping different variations of a word to a single standard form (e.g., "USA" and "U.S.A." $\rightarrow$ "USA").
*   **Lemmatization:** Reducing a word to its **lemma**, which is the base or dictionary form (e.g., "cats" $\rightarrow$ "cat", "is" $\rightarrow$ "be").
*   **Stemming:** A simpler, more aggressive version of lemmatization that simply strips suffixes from the ends of words.
*   **Sentence Segmentation:** Breaking text into individual sentences using cues like periods or exclamation points.


### 3.2 Python Sample: Processing a Corpus
While professional libraries like NLTK or SpaCy are typically used, we can perform basic tokenization and normalization using Python's built-in string methods and regex.

```python
import re

# A sample corpus containing various challenges
corpus = "I'm learning NLP at 2:00 PM! It's fun, isn't it? #learning :) "

# 1. Basic Tokenization (splitting by whitespace)
tokens_basic = corpus.split()
print(f"Basic tokens: {tokens_basic}")

# 2. Advanced Tokenization using Regex (handling punctuation)
# This pattern finds words and emoticons/hashtags
tokens_adv = re.findall(r"[\w']+|[#\w]+|[:\)]+", corpus)
print(f"Advanced tokens: {tokens_adv}")

# 3. Normalization: Lowercasing and stripping whitespace
normalized = [t.lower().strip() for t in tokens_adv]
print(f"Normalized tokens: {normalized}")
```

### 3.3 Practice: Types, Tokens, and Heaps' Law
**Theory Recap:**
*   **Tokens ($N$):** The total number of running words in a corpus.
*   **Types ($|V|$):** The number of distinct words in the vocabulary.
*   **Heaps' Law:** This law states that the vocabulary size ($|V|$) grows significantly faster than the square root of the corpus size ($N$) in words, following the formula $|V| = kN^\beta$.

**Task:** Use the provided text to calculate the number of tokens and types. Then, calculate the "Type-Token Ratio" (TTR), which is $|V|/N$.


```python
import re

# A short sample text
text = "The woodchuck saw the woodchuck near the other woodchuck."

# Tokenize the text (convert to lowercase and remove the trailing period)
# Hint: Use .lower() and .replace(".", "")
clean_text = text.lower().replace(".", "")
tokens = clean_text.split()

# Calculate N (total number of tokens)
N = len(tokens)

# Calculate |V| (number of unique types)
# Hint: Convert the list of tokens into a set() to get unique values
V_size = len(set(tokens))

# Calculate the TTR (Type-Token Ratio)
TTR = V_size / N

print(f"Total Tokens (N): {N}")
print(f"Vocabulary Size (|V|): {V_size}")
print(f"Type-Token Ratio: {TTR:.2f}")
```

## 4. Measuring Similarity and Language Modeling

### 4.1 Theory: Minimum Edit Distance
Natural language processing often requires measuring how similar two strings are. This is used in spelling correction, speech recognition, and coreference resolution.

**Key Concepts**
*   **Minimum Edit Distance:** The minimum number of editing operations (insertions, deletions, substitutions) needed to transform one string into another.
*   **Levenshtein Distance:** The simplest version of this metric where each of the three operations (insertion, deletion, substitution) has a cost of 1.
*   **Alignment:** A visualization of string distance where a correspondence is shown between substrings of the two sequences. For example, the distance between "intention" and "execution" is 5.

### 4.2 Python Sample: Calculating Distance with `nltk`
While we can implement the algorithm manually using a dynamic programming grid, the `nltk` (Natural Language Toolkit) library provides a built-in function to compute Levenshtein distance efficiently.


```python
import nltk

word1 = "intention"
word2 = "execution"

# Compute Levenshtein distance
distance = nltk.edit_distance(word1, word2)

print(f"The edit distance between '{word1}' and '{word2}' is: {distance}")
# Output: 5
```

### 4.4 Theory: N-gram Language Models
A **Language Model** is a probability distribution over sequences of characters or words.

**Key Concepts**
*   **N-gram:** A sequence of written symbols of length $n$.
    *   **Unigram:** 1-gram
    *   **Bigram:** 2-gram
    *   **Trigram:** 3-gram.
*   **Chain Rule of Probability:** Used to decompose the joint probability of an entire sequence $P(w_{1:n})$ into the product of conditional probabilities: $P(w_{1:n}) = \prod_{k=1}^n P(w_k|w_{1:k-1})$.
*   **Markov Assumption:** In practice, we approximate this by assuming the probability of a word depends only on the previous $N-1$ words (e.g., a bigram model looks at only one previous word).

### 4.5 Practice: Estimating N-gram Probabilities
**Task:** Given a small corpus, compute the probability of a specific word given its history using relative frequency counts.

The formula for the bigram probability of "the" given "that" is:
$P(the|that) = \frac{C(that\ the)}{C(that)}$.


```python
# A tiny corpus for training
corpus_text = "its water is so transparent that the water is so transparent that the sun shines"

# Tokenize the corpus
tokens = corpus_text.split()

# Calculate the probability P(the | that)
# 1. Count how many times 'that' appears
count_that = tokens.count("that")

# 2. Count how many times the sequence 'that the' appears
count_that_the = 0
for i in range(len(tokens) - 1):
    if tokens[i] == "that" and tokens[i+1] == "the":
        count_that_the += 1

# 3. Compute probability
prob_the_given_that = count_that_the / count_that if count_that > 0 else 0

print(f"C(that): {count_that}")
print(f"C(that the): {count_that_the}")
print(f"P(the | that) = {prob_the_given_that}")
```

## 5. Small Project—Building a Simple Regex-Based Chatbot (ELIZA)

### 5.1 Theory: Rogerian Psychology and Text Transformation
One of the earliest and most famous examples of a chatbot is **ELIZA** (Weizenbaum, 1966). ELIZA was designed to simulate a **Rogerian psychotherapist**, a style of therapy that involves reflecting a patient's own statements back to them as questions.

The core mechanism of ELIZA is surprisingly simple: it uses **regular expressions** to identify specific patterns in user input and then uses **substitutions** and **capture groups** to transform those statements into responses. For example, if a user says, "I need a holiday," the system identifies the "I need X" pattern and transforms it into "Why do you need a holiday?".

**Key Technical Concepts**
*   **Substitutions:** Replacing a matched pattern with a new string.
*   **Capture Groups:** Using parentheses `( )` to store part of a match in a **numbered register** (like `\1`) so it can be reused in the replacement string.
*   **Backreferences:** Referring back to a captured group during the substitution process.

### 5.2 Python Sample: A Basic Reflection
This snippet demonstrates how to use the `re.sub()` function with a backreference to turn a statement into a question.


```python
import re

user_input = "I need some coffee"

# Define the pattern: Find "I need " followed by anything (.*)
# Wrap (.*) in parentheses to create a capture group
pattern = r"I need (.*)"

# Define the response: \1 refers to whatever was captured in the (.*) group
response = re.sub(pattern, r"Why do you need \1?", user_input)

print(f"User: {user_input}")
print(f"ELIZA: {response}")
# Output: ELIZA: Why do you need some coffee?
```

### 5.3 Project Task: Implementation of "Mini-ELIZA
**Task:** You will complete a Python script that runs a continuous chat loop. You must define at least **three different response patterns** using regular expressions and capture groups.

**Requirements:**
1.  **Requirement 1:** A rule to handle "I am X" (e.g., "I am sad" $\rightarrow$ "Did you come to me because you are sad?").
2.  **Requirement 2:** A rule to handle "I need X" (e.g., "I need help" $\rightarrow$ "What would it mean to you if you got help?").
3.  **Requirement 3:** A "catch-all" or default response for when no patterns match.



```python
import re

def eliza_response(user_text):
    # Rule 1: Handling "I am X"
    if re.search(r"I am (.*)", user_text, re.IGNORECASE):
        return re.sub(r"I am (.*)", r"Did you come to me because you are \1?", user_text, flags=re.IGNORECASE)

    # Rule 2: Handling "I need X"
    pattern_need = r"I need (.*)"
    replacement_need = r"What would it mean to you if you got \1?"
    if re.search(pattern_need, user_text, re.IGNORECASE):
        return re.sub(pattern_need, replacement_need, user_text, flags=re.IGNORECASE)

    # Rule 3: Add one more custom rule (matching "my [something]")
    pattern_custom = r"my (.*)"
    replacement_custom = r"Tell me more about your \1."
    if re.search(pattern_custom, user_text, re.IGNORECASE):
        return re.sub(pattern_custom, replacement_custom, user_text, flags=re.IGNORECASE)

    # Default catch-all response
    return "Please tell me more."

# Simple Chat Loop
print("ELIZA: Hello. How are you feeling today?")
for _ in range(5): # Limit to 5 turns for testing
    user_msg = input("You: ")
    if user_msg.lower() in ["quit", "exit", "bye"]:
        print("ELIZA: Goodbye.")
        break
    print(f"ELIZA: {eliza_response(user_msg)}")
```

## Part 1: The Multivariate Gaussian Distribution

### 1.1. Theory: Covariance Parameterization

A multivariate Gaussian distribution for a set of continuous variables $X={X_1,…,X_n}$ is defined by a mean vector $μ$ and a covariance matrix $Σ$.
- Mean Vector ($μ$): Represents the expected value of each variable.
- Covariance Matrix ($Σ$): $Σ_{i,i}$ is the variance of $X_i$, and $Σ_{i,j}$ is the covariance between $X_i$ and $X_j$.
_ Positive Definite Requirement: For the density to be well-defined, Σ must be positive definite (for any $x ≠ 0, x^T Σx>0$), which ensures it is non-singular and has a non-zero determinant.

## Part 2: The Information Form

### 2.1 Theory: Precision and Potentials

Because positive definite matrices are invertible, we can utilize an alternative parameterization called the Information Form:
- Information Matrix ($J$): The inverse of the covariance matrix ($J=Σ^−1$), also known as the precision matrix.
- Potential Vector ($h$): Defined as $h=Jμ$.

The density in this form is proportional to $exp[−\frac{1}{2}x^TJx+(Jμ)^T x]$.

## Part 3: Independencies in Gaussians


### 3.1. Theory: Marginal vs. Conditional Independence


Independencies in multivariate Gaussians are encoded directly in the parameters:
- Marginal Independence: $X_i$ and $X_j$ are independent if and only if $Σ_{i,j} = 0$.
- Conditional Independence: $X_i$ and $X_j$ are conditionally independent given all other variables if and only if $J_{i,j} = 0$.

### 1.1 From Directed to Undirected
**Theory:** In your previous labs, you worked with **Bayesian Networks**, which are **directed acyclic graphs (DAGs)**. While Bayesian Networks are excellent for representing causal relationships, they cannot easily represent symmetrical dependencies (like the relationship between two neighboring pixels in an image).

**Markov Networks** solve this by using **undirected edges**. Key characteristics include:
*   **Pairwise Markov Networks:** The simplest form where edges represent direct dependencies between two variables.

*   **Factorization:** The joint distribution is represented as a product of factors over the cliques of the graph.

## Part 2: The General Gibbs Distribution

Following our exploration of factorization in Part 1, we now look at how to define a complete **Joint Probability Distribution** for an undirected model. In the course syllabus, this is referred to as the **General Gibb distribution**.

### 2.1 The Partition Function ($Z$)
**Theory:** In a Bayesian Network, the product of CPDs is guaranteed to sum to 1. In a Markov Network, the product of factors ($\phi$) results in an **unnormalized measure** $\tilde{P}(X)$. To convert this into a valid probability distribution $P(X)$, we must divide by a normalizing constant called the **Partition Function ($Z$)**.

The formula for a Gibbs distribution is:
$$P(X_1, ..., X_n) = \frac{1}{Z} \prod_{i} \phi_i(C_i)$$

Where:
*   $\phi_i(C_i)$ are the factors over cliques in the graph.
*   $Z = \sum_{X} \prod_{i} \phi_i(C_i)$ is the sum of the unnormalized weights across **all possible states** of the network.

**Why is $Z$ important?**
$Z$ ensures that $\sum P(X) = 1$. However, calculating $Z$ is often the most computationally expensive part of inference because it requires summing over an exponential number of possible state combinations.

### 2.2 Example: Unnormalized Probability
Let's use our "Study Group" triangular network (A-B-C) from Part 1. We assume each edge has a factor that rewards students for having the same "Study Status" (1 for studying, 0 for not).

```python
# Define factors for edges (A-B), (B-C), (C-A)
# For simplicity, we'll use the same affinity factor for all
phi = {(0,0): 5, (0,1): 1, (1,0): 1, (1,1): 5}

def get_unnormalized_weight(state_a, state_b, state_c):
    # P_tilde = phi(A,B) * phi(B,C) * phi(C,A)
    w_ab = phi[(state_a, state_b)]
    w_bc = phi[(state_b, state_c)]
    w_ca = phi[(state_c, state_a)]

    return w_ab * w_bc * w_ca

# Calculate unnormalized weight for the state where everyone is studying (1,1,1)
weight_111 = get_unnormalized_weight(1, 1, 1)
print(f"Unnormalized weight for state (1,1,1): {weight_111}")
# Output: 5 * 5 * 5 = 125
```

## Part 3: Independence and the Markov Blanket

In this section, we move from the numerical representation of probabilities to the **structural properties** of the network. We will explore how the lack of edges in a Markov Network signifies conditional independence.




### 3.1  Separation and Markov Properties
**Theory:** In Bayesian Networks, we used **d-separation** (which involved complex rules about head-to-head meeting and directions). In **Markov Networks**, independence is much simpler because the edges are undirected:

*   **Global Markov Property (Separation):** Two sets of nodes $X$ and $Y$ are conditionally independent given a set of observed nodes $Z$ (written as $X \perp Y \mid Z$) if **every path** between any node in $X$ and any node in $Y$ passes through at least one node in $Z$.
*   **The Markov Blanket:** In an undirected graph, the Markov Blanket of a node $X$ is simply the set of its **immediate neighbors**. Given its neighbors, $X$ is independent of all other nodes in the network.

### 3.2 Example: Finding the Markov Blanket
Using the `networkx` library, we can easily programmatically identify these dependencies. This is a crucial skill for simplifying large-scale inference problems.



```python
import networkx as nx

# Create a more complex "Social Network" Markov model
# Edges represent mutual influence between friends
social_net = nx.Graph()
social_net.add_edges_from([
    ('Alice', 'Bob'), ('Bob', 'Charlie'), ('Alice', 'David'),
    ('David', 'Eve'), ('Charlie', 'Eve'), ('Eve', 'Frank')
])

def get_markov_blanket(graph, node):
    # In a Markov Network, the blanket is just the neighbors
    return list(graph.neighbors(node))

blanket_eve = get_markov_blanket(social_net, 'Eve')
print(f"The Markov Blanket for Eve is: {blanket_eve}")
# Output: ['David', 'Charlie', 'Frank']
# This means if we know the state of David, Charlie, and Frank,
# Eve's state is independent of Alice and Bob.
```

## Part 4: Log-Linear Models

In this final theoretical section before the project, we explore **Log-Linear Models**. As outlined in the course syllabus (Chapter 2, Section f), log-linear models provide a powerful way to represent Markov Networks using features and weights.

### **4.1 Features and Energy**
**Theory:** Representing a distribution as a product of factors can be computationally difficult due to the very small numbers involved (underflow). Instead, we often work in **log-space**.

A factor $\phi(D)$ can be represented as:
$$\phi(D) = \exp(-w \cdot f(D))$$

Where:
*   $f(D)$ is a **feature function** that returns 1 if a certain condition is met and 0 otherwise.
*   $w$ is a **weight** representing the strength/importance of that feature.

By taking the log of the Gibbs distribution, the product of factors becomes a **weighted sum of features**, also known as the **Energy Function**:
$$Energy(X) = \sum_{i} w_i f_i(D_i)$$
The higher the energy, the lower the probability ($P(X) \propto \exp(-Energy)$).

### 4.2 Visualization: Energy vs. Probability
In log-linear models, we don't just look at tables; we look at "energy landscapes." A state with many violated constraints has high energy and low probability.


```python
import numpy as np
import matplotlib.pyplot as plt

# Visualize the relationship: P(X) = exp(-Energy)
energies = np.linspace(0, 10, 100)
probabilities = np.exp(-energies)

plt.plot(energies, probabilities)
plt.title("Relationship between Energy and Probability")
plt.xlabel("Energy (Weighted Sum of Features)")
plt.ylabel("Unnormalized Probability")
plt.grid(True)
plt.show()
```

### 5.1 Theory: Markov Random Fields (MRFs)
A **Markov Random Field (MRF)** is a specific type of Markov Network where the nodes are arranged in a grid. In image processing, we treat each pixel as a node in the graph.

To denoise an image, we model two types of interactions using factors:
1.  **Observation Evidence (Node Potential):** We want the "true" pixel value to be similar to the "noisy" observed value.
2.  **Spatial Consistency (Edge Potential):** We expect neighboring pixels to have similar values (images are generally "smooth").

We combine these into an **Energy Function** (from Part 4). The goal of the project is to find a configuration of pixels that minimizes this energy.

### 5.2 Visualization: The Grid Model
Imagine a grid where every pixel $X_i$ is connected to its four neighbors (Up, Down, Left, Right).
*   **Nodes:** The hidden "true" pixel values we want to find.
*   **Edges:** Represent the "Agreement Feature" ($w_{agree}$) ensuring smoothness.
*   **Evidence:** Each hidden node is also connected to the observed noisy pixel $Y_i$.

# Part 1: Foundations of Probability Theory

In this first part, we will review the basic building blocks of probability theory, including event spaces, conditional probability, and the power of Bayes' Rule.

## 1.1. Event Spaces and Probability Distributions

**Theory**

Probability theory provides a formal foundation for discussing uncertain events
* Sample Space $(Ω)$: The set of all possible outcomes. For a single die, $Ω=\{1,2,3,4,5,6\}$
* Events $(S)$: A set of measurable outcomes. Each event α∈S is a subset of $Ω$
* Probability Distribution $(P)$: A mapping from events to real values that must satisfy:
  1.   $P(α)≥0$ (Non-negativity)
  2.   $P(Ω)=1$ (The trivial event has maximal probability)
  3.   If $α,β$ are disjoint $(α∩β=∅)$, then $P(α∪β)=P(α)+P(β)$



In Python, we can represent a discrete probability distribution using a **dictionary** where keys are outcomes and values are their probabilities.

```python
# Representing a fair 6-sided die
die_dist = {}

def get_event_probability(event, distribution):
    """Calculates probability of a subset of outcomes."""
    return sum(distribution[outcome] for outcome in event)

# Probability of rolling an odd number: {1, 3, 5}
odd_event = {1, 3, 5}
print(f"P(Odd) = {get_event_probability(odd_event, die_dist):.2f}")
```

## 1.2. Conditional Probability


**Theory**

Conditional probability allows us to reason about how new evidence changes our beliefs. The probability of event $β$ occurring given that we know event $α$ has occurred is defined as:

$P(β∣α)= \frac{P(α∩β)}{P(α)}$​

This is the relative proportion of outcomes satisfying $β$ among those that already satisfy $α$.


## 1.3. Bayes' Rule


**Theory**

Derived from the definition of conditional probability, Bayes' Rule is a critical tool for "inverting" conditional probabilities

$P(α∣β)= \frac{P(β∣α)P(α)}{P(β)}$

This allows us to compute the posterior $P(α∣β)$ using the likelihood $P(β∣α)$ and the prior $P(α)$.

In Part 1, we focused on events. However, in complex AI systems, it is more natural to consider attributes of outcomes, such as a student's "Grade" or "Intelligence". This part introduces the formal machinery to handle these attributes.

## 2.1. What is a Random Variable?


**Theory**

A random variable is defined by a function that associates a value with each outcome in the sample space $Ω$.
* Discrete Random Variables: Take on a finite set of values (e.g., $Grade ∈ \{A,B,C\}$).
* Continuous Random Variables: Take on real values (e.g., a student's height).

We use the notation $P(X=x)$ to denote the probability that the random variable $X$ takes the value $x$. For a binary-valued variable $X∈\{0,1\}$, the distribution is often called a Bernoulli distribution.

We can represent a discrete random variable's distribution using a dictionary where the keys are the possible values $(Val(X))$ and the values are their probabilities.

```python
# A Bernoulli distribution for 'Intelligence'
# I=1 (High), I=0 (Low)
intelligence_dist = {1: 0.3, 0: 0.7}

def check_valid_distribution(dist):
    """Ensures the sum of probabilities is 1."""
    return sum(dist.values()) == 1.0

print(f"Is valid: {check_valid_distribution(intelligence_dist)}")
```

## 2.2. Joint and Marginal Distributions

When we are interested in multiple attributes simultaneously, we use a Joint Distribution. For two variables $X$ and $Y$, the joint distribution $P(X,Y)$ assigns probabilities to all combinations of their values.

**Marginalization** is the process of extracting the distribution of a single variable from a joint distribution by "summing out" the other variables:

$P(x)=\sum_{y∈Val(Y)}^{} P(x,y)$

As the number of variables increases, the number of parameters grows exponentially. For $n$ binary variables, a joint distribution requires $2^
n − 1$ numbers, which becomes computationally unmanageable.

## 2.3. Conditional Probability for Random Variables


**Theory**

The concept of conditional probability extends naturally to random variables. The distribution $P(X∣Y=y)$ represents our updated belief about $X$ after observing that $Y$ has a specific value $y$.

# Part 3: Independence and Graph Theory Basics

In this part, we explore how to simplify complex distributions by identifying **independence** and how to visualize these relationships using **Graph Theory**.

## 3.1. Independence and Conditional Independence


**Theory**

Independence: Two events $α$ and $β$ are independent if learning that $β$ occurred does not change our belief in $α$. Formally: $P(α∣β)=P(α)$ or $P(α∩β)=P(α)P(β)$.

Conditional Independence: We say $α$ is conditionally independent of $β$ given $γ$ if $P(α∣β∩γ)=P(α∣γ)$. This is the core concept that allows Bayesian Networks to represent joint distributions compactly.

We can verify independence by checking if the joint probability equals the product of the marginals.

```python
def are_independent(joint_dist, p_x, p_y, x_val, y_val, epsilon=1e-9):
    """
    Checks if P(X=x, Y=y) == P(X=x) * P(Y=y)
    epsilon is used to handle floating point comparisons.
    """
    p_joint = joint_dist[y_val][x_val]
    return abs(p_joint - (p_x * p_y)) < epsilon
```

## 3.2. Graph Theory Basics: Nodes and Edges


**Theory**

A graph is a data structure $K$ consisting of:
1. **Nodes** $(X)$: Representing our random variables $\{X_1,…,X_n\}$.
2. **Edges** $(E)$: Representing relationships. A directed edge $X_i→X_j$ implies $X_i$ is a parent of $X_j(Pa_{X_j})$.
3. **Directed Acyclic Graph (DAG)**: A graph with only directed edges and **no cycles** (you cannot follow a path from a node back to itself). Bayesian Networks are built on DAGs.

The most common library for graph structures in Python is networkx.

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a Directed Graph (DAG)
dag = nx.DiGraph()

# Add nodes (Variables)
dag.add_nodes_from(['Intelligence', 'Grade', 'SAT'])

# Add edges (Dependencies)
# Intelligence influences both Grade and SAT score
dag.add_edges_from([('Intelligence', 'Grade'), ('Intelligence', 'SAT')])

# Visualize
nx.draw(dag, with_labels=True, node_color='lightblue', node_size=2000)
plt.show()

# Find parents of a node
print(f"Parents of Grade: {list(dag.predecessors('Grade'))}")
```

# Part 4: Bayesian Networks and Factorization


In this final part, we bring everything together. We have seen that explicit joint distributions are **computationally unmanageable** because they require $2^n−1$ parameters for $n$ binary variables. **Bayesian Networks (BNs)** solve this by using the graph structure to represent the distribution in a **compact**, **factorized way**.

## 4.1. Formal Definition of a Bayesian Network

**Theory**

A Bayesian Network is a pair $B=(G,P)$ where:

1. $G$ is a Directed Acyclic Graph (DAG) whose nodes represent random variables
$X_1,…,X_n$.
2. $P$ is a probability distribution that **factorizes** according to $G$.

The core idea is the **Local Independence Assumption**: each node
$X_i$ is conditionally independent of its non-descendants given its parents
$(Pa_{X_i})$.

## 4.2. The Chain Rule for Bayesian Networks


**Theory**

The most important result of this factorization is the **Chain Rule for BNs**. It allows us to calculate the probability of any complete assignment of values (the joint probability) by multiplying the local **Conditional Probability Distributions (CPDs)** for each node:

$P(X_1,…,X_n) = \prod_{i=1}^{n} P(X_i|Pa_{X_i})$
